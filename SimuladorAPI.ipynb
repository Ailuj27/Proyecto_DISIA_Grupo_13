```python
import requests
import threading
import time

URL = "http://localhost:8000/health" 

def bombardear_api():
    while True:
        try:
            requests.get(URL)
        except:
            pass
        time.sleep(0.05) 

print("🚀 Iniciando tráfico masivo hacia el endpoint /health...")
print("⚠️ Ve a Grafana. En 10-20 segundos verás subir la gráfica de 'Peticiones por Segundo'.")
print("Pulsa Ctrl+C en esta terminal para detener el simulador.")

# Lanzamos 10 hilos simultáneos
for _ in range(10):
    hilo = threading.Thread(target=bombardear_api)
    hilo.daemon = True
    hilo.start()

try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("\n🛑 Simulación detenida.")
```
